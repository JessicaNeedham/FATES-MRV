#### Libraries

#### make a sequence of seed rain values and generate parameter files

```python
seed_rains = np.linspace(0.018, 0.18, 10)
seed_germinations = np.linspace(0.001, 0.2, 10)
print(seed_rains)
print(seed_germinations)
```

```python
# input parameter file name
inputf = '/global/homes/j/jneedham/FATES-MRV/param_files/v9/fates_params_PA_nat.nc'
organ = 1
counter = 1

for i in range(0,len(seed_rains)): 

    for j in range(0,len(seed_rains)): 
    
        fout = '/global/homes/j/jneedham/FATES-MRV/param_files/v9/fates_params_PA_nat_seedrain_{0}.nc'.format(counter)

        shutil.copyfile(inputf, fout)  

        var = 'fates_recruit_seed_supplement'
        pft = 1
        val = seed_rains[i]
        mp.main(var = var, pft = pft, fin=fout, val = val, 
                fout = fout, O = 1, organ = organ)
        pft = 2
        val = seed_rains[i] * 0.9
        mp.main(var = var, pft = pft, fin=fout, val = val, 
                fout = fout, O = 1, organ = organ)

        var = 'fates_recruit_seed_germination_rate'
        pft = 1
        val = seed_germinations[j]
        mp.main(var = var, pft = pft, fin=fout, val = val, 
                fout = fout, O = 1, organ = organ)
        pft = 2
        val = seed_germinations[j] * 0.9
        mp.main(var = var, pft = pft, fin=fout, val = val, 
                fout = fout, O = 1, organ = organ)

        counter = counter + 1
```
