### AD SPIN UP

```python
#fates = '/pscratch/sd/j/jneedham/fates-mrv-runs/runs/v8/PA_AD_spinup_v8.Ea8fbc2fc28-F1088cf26.2025-07-07/run/PA_AD_spinup_v8.Ea8fbc2fc28-F1088cf26.2025-07-07.sofar.nc'
fates = '/pscratch/sd/j/jneedham/fates-mrv-runs/runs/v8/PA_AD_spinup_test17.E45487965d0-Fb8b74da5.2025-07-17/run/'
fates = xr.open_dataset(fates, decode_times=False)
```

```python
time = fates.variables['time'] / 365.
```

```python
AGB = fates.FATES_VEGC_ABOVEGROUND_SZPF  * 10000 * 0.001 * 2
AGB = AGB.sum(dim='fates_levscpf')

fig1, ((f1ax0)) = plt.subplots(nrows=1,ncols=1, figsize=(12,6))

f1ax0.plot(time, AGB, color='green', label = 'FATES', alpha = 0.4)    
f1ax0.set_title(r'AGB - AD spinup', fontsize = 19)
f1ax0.set_ylabel(r'AGB (Mg C ha$^{-1}$)', fontsize = 19)
f1ax0.set_xlabel(r'Time (years)', fontsize = 19)    
```
